```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
   "Region": ["Riyadh", "Makkah", "Madinah", "Eastern", "Asir"],
   "Adoption": [85, 72, 91, 65, 78]
}

df = pd.DataFrame(data)
```

```python
# sorting the data by adoption
df_sorted = df.sort_values("Adoption")

# setting bar chart
plt.figure(figsize=(8,5))
plt.bar(df_sorted["Region"], df_sorted["Adoption"], color='skyblue', edgecolor='black')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Sorted")
plt.xticks(rotation=45)

plt.show()
```

```python
# setting bar chart
plt.figure(figsize=(8,5))
plt.bar(df["Region"], df["Adoption"], color='skyblue', edgecolor='black')
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Unsorted")
plt.xticks(rotation=45)

plt.show()
```

```python
# highlighting madinah
explode = (0, 0, 0.1, 0, 0)

# setting pie chart
plt.pie(
   df["Adoption"],
   labels=df["Region"],
   autopct="%1.0f%%",
   colors=sns.color_palette("Set2"),
   explode=explode,
   wedgeprops={'edgecolor': 'black'}
)

plt.title("Digital Adoption by Region — Pie")

plt.show()
```

```python
# setting the bubble chart
plt.scatter(
   df["Region"],
   df["Adoption"],
   s=df["Adoption"] * 20,
   alpha=0.7,
   edgecolors='black'
)
plt.grid(True,linestyle='--', alpha=0.7)

plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.title("Digital Adoption by Region — Bubble")
plt.xticks(rotation=45)

plt.show()
```

```python
# sorting data
df_sorted = df.sort_values("Adoption")

# setting heatmap
plt.figure(figsize=(10,2))
sns.heatmap(
   [df_sorted["Adoption"]],
   annot=True,
   xticklabels=df_sorted["Region"],
   yticklabels=["Adoption"],
   cmap="Blues"
)

plt.title("Digital Adoption by Region — Heatmap")
plt.show()
```
