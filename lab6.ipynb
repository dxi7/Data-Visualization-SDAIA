```python
BIG_IDEA = "Prioritize lagging regions to improve adoption."

STORYBOARD = [
    {"arc_role": "BLUF", "message": "Prioritize investment in lagging regions."},
    {"arc_role": "Situation", "message": "Adoption varies across regions."},
    {"arc_role": "Complication", "message": "Three regions are below the 65% target."},
    {"arc_role": "Evidence", "message": "East has the largest adoption gap (17 pts)."},
    {"arc_role": "Options", "message": "Broad investment vs. targeted regional investment."},
    {"arc_role": "Resolution", "message": "Target investment at lagging regions."},
    {"arc_role": "Ask", "message": "Approve the targeted investment approach."}
]

TIME_BOX = {
    "BLUF": 30, "Situation": 45, "Complication": 60,
    "Evidence": 75, "Options": 90, "Resolution": 60, "Ask": 30
}

def build_notes(storyboard):
    total = 0
    print("="*65)
    print(f" OPEN (Say the Ask): {BIG_IDEA}")
    print("="*65 + "\n")

    for s in storyboard:
        role = s["arc_role"]
        message = s["message"]
        secs = TIME_BOX.get(role, 45)
        total += secs

        minutes = total // 60
        seconds = total % 60

        print(f"⏱️ [{minutes}:{seconds:02d}] {role.upper()}")
        print(f"   ➤ Message: {message}")
        print(f"   ➤ Action: Orient axes, point at one mark (Do NOT read it).\n")

    assert total <= 7*60, f"Over by {total-420}s — cut."

    print("="*65)
    print("🏁 CLOSE (Ask AGAIN): Approve targeted investment; set 90-day check.")
    print("="*65)

build_notes(STORYBOARD)
```

```python
QA = [
    {"q": "Aren't those regions just small? (Hostile)",
     "answer": "Per-capita, not absolute (See Appendix A2).",
     "bridge": "So targeted spend closes the real gap."},

    {"q": "Adoption looks high vs finance? (Data-wrong)",
     "answer": "Unique users, not a column average (See Appendix A5).",
     "bridge": "It holds at finance's lower figure."},

    {"q": "Why not spend it on the app? (Off-topic)",
     "answer": "Channel infra is option 2; slower.",
     "bridge": "Region-targeting is the fastest way to 65%."}
]

def drill(qa):
    print("="*60)
    print(" Q&A PREP MATRIX (Acknowledge -> Answer -> Bridge)")
    print("="*60 + "\n")

    for i, item in enumerate(qa, 1):
        print(f" Q{i}: {item['q']}")
        print(f"    Answer: {item['answer']}")
        print(f"    Bridge: ... {item['bridge']}\n")

drill(QA)
```
