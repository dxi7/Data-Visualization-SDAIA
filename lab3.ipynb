```python
#before
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
   "Department": ["HR", "Finance", "IT", "Sales","Dev"],
   "Absence": [5, 8, 12, 7,9]
})

plt.bar(df["Department"], df["Absence"],
       color=["red", "green", "blue", "purple"])
plt.title("Employee Absence")
plt.xlabel("Department")
plt.ylabel("Absence %")
plt.grid(True)
plt.legend(["Absence"])
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"],edgecolor='black')

plt.title("Employee Absence by Department")
plt.xlabel("Department")
plt.ylabel("Absence (%)")

plt.show()
```

```python
#before
plt.figure(figsize=(8,5))
plt.bar(df["Department"], df["Absence"], color="steelblue")
plt.title("IT Has the Highest Absence — 12%")
plt.xlabel("Department")
plt.ylabel("Absence (%)")
plt.grid(False)

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))
plt.show()
```

```python
plt.bar(df["Department"], df["Absence"],edgecolor='black')
plt.xlabel("Department")
plt.ylabel("Absence (%)")

plt.axhline(df["Absence"].mean(),color='red',linestyle="--",label=f"Avg of absence,{df["Absence"].mean():.0f}")
plt.grid(axis='y',alpha=0.7,linestyle="--")

ax = plt.gca()
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.annotate("Highest",
            xy=("IT", 12),
            xytext=("Finance", 13),
            arrowprops=dict(arrowstyle="->"))

plt.legend()
plt.show()
```
