```python
storyboard = [
    ["BLUF", "Prioritize investment in lagging regions."],
    ["Situation", "Adoption varies across regions."],
    ["Complication", "Some regions are below the 65% target."],
    ["Evidence", "The largest gaps are concentrated in lagging regions."],
    ["Options", "Broad investment or targeted regional investment."],
    ["Recommendation", "Target investment at lagging regions."],
    ["Ask", "Approve the targeted investment approach."]
]

for slide, message in enumerate(storyboard, start=1):
    print(f"Slide {slide}: {message[0]}")
    print("Message:", message[1])
    print()

def audit(storyboard):
    checks = {
        "Exactly 7 scenes": len(storyboard) == 7,
        "Starts with BLUF": storyboard[0][0] == "BLUF",
        "Ends with Ask": storyboard[-1][0] == "Ask",
        "Every scene has a message": all(
            message.strip() for _, message in storyboard
        )
    }

    for check, result in checks.items():
        print(f"{'PASS' if result else 'FAIL'} — {check}")

    return all(checks.values())

audit(storyboard)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Region": ["North", "South", "East", "West"],
    "Adoption": [72, 61, 48, 55]
})

target = 65
print(df)
```

```python
plt.figure(figsize=(8, 5))

bars = plt.bar(df["Region"], df["Adoption"],edgecolor='black')

plt.title("Adoption Varies Across Regions", fontsize=12)
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.ylim(0, 100)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 2, f"{int(yval)}%", ha='center', va='bottom')

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

colors = ['gray' if val >= target else '#1f77b4' for val in df["Adoption"]]

bars = plt.bar(df["Region"], df["Adoption"], color=colors, edgecolor='black')

plt.axhline(target, linestyle="--", color="black",label="Target = 65%")

plt.title("Three Regions Are Below the 65% Target")
plt.xlabel("Region")
plt.ylabel("Adoption (%)")
plt.ylim(0, 100)
plt.legend()

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 2, f"{int(yval)}%", ha='center', va='bottom', fontweight='bold')

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.show()
```

```python
df["Gap"] = target - df["Adoption"]

plt.figure(figsize=(8, 5))

colors = ['lightgray', 'lightgray', '#1f77b4', 'lightgray']
bars = plt.bar(df["Region"], df["Gap"], color=colors, edgecolor='black', alpha=0.9)

plt.axhline(0, color='black', linewidth=1.2)

plt.title("East Has the Largest Adoption Gap", fontsize=14, fontweight='bold')
plt.xlabel("Region", fontsize=12)
plt.ylabel("Gap to Target", fontsize=12)

plt.ylim(df["Gap"].min() - 5, df["Gap"].max() + 5)

for bar in bars:
    yval = bar.get_height()
    if yval > 0:
        plt.text(bar.get_x() + bar.get_width()/2, yval + 1, f"+{int(yval)}", ha='center', va='bottom', fontweight='bold')
    else:
        plt.text(bar.get_x() + bar.get_width()/2, yval - 1, f"{int(yval)}", ha='center', va='top', fontweight='bold')

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)

plt.show()
```
